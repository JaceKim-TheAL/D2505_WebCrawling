````python

# driver = webdriver.Chrome("C:/Python/driver/chromedriver-win64/chromedriver.exe")
driver = webdriver.Chrome()
````

````python
driver.get("https://www.naver.com/")
````

````python
# Naver login 클릭
driver.find_element(By.XPATH, '//*[@id="account"]/div/a').click()
````

````python
# 이전 페이지로 이동
driver.back()  
````

````python
# Naver login 클릭
login_btn = driver.find_element(By.CSS_SELECTOR, "#account > div > a")
login_btn.click()
````

````python
# 아이디 입력
naver_id = driver.find_element(By.CSS_SELECTOR, "#id")
naver_id.send_keys(config.NAVER_ID)
````

````python
# 비밀번호 입력
naver_pw = driver.find_element(By.CSS_SELECTOR, "#pw")
naver_pw.send_keys(config.NAVER_PW)
````

````python
# 로그인인 클릭
login_btn = driver.find_element(By.CSS_SELECTOR, "#log\.login")
# login_btn.click()
````
