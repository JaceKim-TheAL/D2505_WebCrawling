```python
target_url_pages = "https://startcoding.pythonanywhere.com/basic?page={pg}"

p_category_arr = []
p_name_arr = []
p_link_arr = []
p_price_arr = []

for i in range(1, 5):
    target_url = target_url_pages.format(pg=i)
    response = requests.get(target_url)
    html = response.text
    soup = BeautifulSoup(html, 'html.parser')

    print(target_url)
    print("-" * 50)
    items = soup.select(".product")

    for item in items:
        p_category = item.select_one(".product-category").get_text()

        item_name  = item.select_one(".product-name")   
        p_name = item_name.get_text()
        p_link = item_name.select_one("a")["href"]

        p_price = item.select_one(".product-price").get_text()
        p_price = p_price.split("원")[0]
        p_price = p_price.replace(",", "")
        p_price = p_price.strip() 
        p_price = int( p_price )

        # print(f"카테고리 : {p_category}, 상품명 : {p_name}, 가격 : {p_price} 원, 상세페이지 : {p_link}")
        p_category_arr.append(p_category)
        p_name_arr.append(p_name)
        p_link_arr.append(p_link)
        p_price_arr.append(p_price)
```

```python
len(p_category_arr), len(p_name_arr), len(p_link_arr), len(p_price_arr)
```

```python
data = { '카테고리':p_category_arr, '상품명':p_name_arr, '상세페이지':p_link_arr, '가격':p_price_arr }
df = pd.DataFrame(data) 
df.head(10)
```

```python
df = pd.DataFrame(data, columns=[ '카테고리', '상품명', '가격', '상세페이지'])
df.tail(5)
```
