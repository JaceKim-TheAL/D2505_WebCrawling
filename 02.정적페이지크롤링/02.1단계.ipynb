### 1단계 : 한개의 상품 크롤링
> 수집할 정보 : 카테고리, 상품명, 상세페이지링크, 가격

```python
import requests
from bs4 import BeautifulSoup
```

```python
target_url = "https://startcoding.pythonanywhere.com/basic"

response = requests.get(target_url)
html = response.text
soup = BeautifulSoup(html, 'html.parser')
soup   
```

```python
product = soup.select_one(".product")
product  
```

```python
p_category = product.select_one(".product-category").get_text() 
p_category
```

```python
p_name = product.select_one(".product-name").get_text() 
p_name
```

```python
p_link = product.select_one("#product-container > div:nth-child(1) > div > div.product-body > h3 > a")
p_link = p_link["href"]
p_link
```

```python
p_price = product.select_one(".product-price").get_text()
p_price = p_price.strip() 
p_price = p_price.replace(",", "")
p_price = p_price.replace("원", "")
p_price = int( p_price )
p_price
```
