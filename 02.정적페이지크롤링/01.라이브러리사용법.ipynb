````python
import requests
from bs4 import BeautifulSoup
````

````python
target_url = "https://startcoding.pythonanywhere.com/basic"

response = requests.get(target_url)
html = response.text
soup = BeautifulSoup(html, 'html.parser')
soup.select_one("div")
````

````python
print(soup.select_one("div").prettify())
````

````python
# 좌측상단 [스타트코딩]을 선택자로 호출
soup.select_one(".brand-name")
````

````python
soup.select_one(".brand-name").text
````

````python
soup.select_one(".brand-name").attrs
````

````python
soup.select_one(".brand-name").attrs['target']
````
